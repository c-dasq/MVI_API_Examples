# 17. Clone a dataset

Create a copy of an existing dataset

Before running: Set DATASET_ID; this creates a new dataset

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
NEW_DATASET_NAME = 'training_clone'

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

payload = {"action": "clone", "name": NEW_DATASET_NAME}
response = requests.post(f"{API_URL.rstrip('/')}/datasets/{DATASET_ID}/action", headers=headers, json=payload, verify=VERIFY_SSL, timeout=120)
response.raise_for_status()
print(response.json())
```
