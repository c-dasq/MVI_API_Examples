# 06. Get file metadata

Read user metadata for one dataset file

Before running: Set DATASET_ID and FILE_ID

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
FILE_ID = '' # The file_id must be retrieved from 05_list_dataset_files

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

response = requests.get(f"{API_URL.rstrip('/')}/datasets/{DATASET_ID}/files/{FILE_ID}/user-metadata", headers=headers, verify=VERIFY_SSL, timeout=30)
response.raise_for_status()
print(response.json())
```
