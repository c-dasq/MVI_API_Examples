# 19. Infer a video frame

Send the first frame from a local video to a deployed model

Before running: Set MODEL_ID and VIDEO_PATH; this notebook uses only the first frame

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
MODEL_ID = '' # The model_id can be found in the URL for that model in the MVI UI
VIDEO_PATH = './example.mp4'

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

import cv2
import tempfile
from pathlib import Path

video = cv2.VideoCapture(VIDEO_PATH)
try:
    success, frame = video.read()
finally:
    video.release()
if not success:
    raise ValueError("Could not read the first video frame")

# Encode the frame as a JPEG before inference
success, image_data = cv2.imencode(".jpg", frame)
if not success:
    raise ValueError("Could not encode the frame")
files = {"files": ("frame.jpg", image_data.tobytes(), "image/jpeg")}
response = requests.post(f"{API_URL.rstrip('/')}/dlapis/{MODEL_ID}", headers=headers, files=files, verify=VERIFY_SSL, timeout=120)
response.raise_for_status()
print(response.json())
```
