# 05. List dataset files

List the files in one dataset

Before running: Set DATASET_ID to an existing dataset ID

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

response = requests.get(f"{API_URL.rstrip('/')}/datasets/{DATASET_ID}/files", headers=headers, verify=VERIFY_SSL, timeout=30)
response.raise_for_status()
for file in response.json():
    print(file.get("_id"), file.get("original_file_name"))
```
