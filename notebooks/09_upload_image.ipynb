# 09. Upload an image

Upload a local image without labels

Before running: Set DATASET_ID and IMAGE_PATH; this uploads a file

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
IMAGE_PATH = './example.jpg'

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

from pathlib import Path

# The multipart field name is "files"
with open(IMAGE_PATH, "rb") as image_file:
    files = {"files": (Path(IMAGE_PATH).name, image_file)}
    response = requests.post(f"{API_URL.rstrip('/')}/datasets/{DATASET_ID}/files", headers=headers, files=files, verify=VERIFY_SSL, timeout=120)
response.raise_for_status()
print(response.json())
```
