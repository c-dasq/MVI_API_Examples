# 20. Copy one file to another dataset

Download one file and upload it to a different dataset

Before running: Set source and target dataset IDs and the source file ID; this copies file content only, not labels

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
SOURCE_DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
TARGET_DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
FILE_ID = '' # The file_id must be retrieved from 05_list_dataset_files
FILE_NAME = './copied_image.jpg'

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

# Download the original file
source_url = f"{API_URL.rstrip('/')}/datasets/{SOURCE_DATASET_ID}/files/{FILE_ID}/download"
download = requests.get(source_url, headers=headers, verify=VERIFY_SSL, timeout=120)
download.raise_for_status()

# Upload the file to the target dataset
files = {"files": (FILE_NAME, download.content)}
target_url = f"{API_URL.rstrip('/')}/datasets/{TARGET_DATASET_ID}/files"
response = requests.post(target_url, headers=headers, files=files, verify=VERIFY_SSL, timeout=120)
response.raise_for_status()
print(response.json())
```
