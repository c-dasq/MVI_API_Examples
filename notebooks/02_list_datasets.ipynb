# 02. List datasets

Get the datasets available to the API key

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

response = requests.get(f"{API_URL.rstrip('/')}/datasets", headers=headers, verify=VERIFY_SSL, timeout=30)
response.raise_for_status()
for dataset in response.json():
    print(dataset.get("_id"), dataset.get("name"))
```
