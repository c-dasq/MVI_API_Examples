# 07. Add file metadata

Add one metadata field to a file

Before running: Set DATASET_ID and FILE_ID; this changes file metadata

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
FILE_ID = '' # The file_id must be retrieved from 05_list_dataset_files
METADATA_KEY = 'example' # several pairs of keys and values can be added
METADATA_VALUE = 'training' # several pairs of keys and values can be added

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

# This adds or updates user metadata on the file
metadata = {METADATA_KEY: METADATA_VALUE}
response = requests.post(f"{API_URL.rstrip('/')}/datasets/{DATASET_ID}/files/{FILE_ID}/user-metadata", headers=headers, json=metadata, verify=VERIFY_SSL, timeout=30)
response.raise_for_status()
print(response.json())
```
