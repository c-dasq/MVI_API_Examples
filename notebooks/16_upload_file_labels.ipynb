# 16. Add file labels

Add example bounding-box labels to a dataset file

Before running: Set DATASET_ID and FILE_ID; this changes file labels

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
FILE_ID = '' # The file_id must be retrieved from 05_list_dataset_files

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

# Replace the example label and coordinates for your image
labels = {"labels": [{"name": "object", "bndbox": {"xmin": 10, "ymin": 10, "xmax": 100, "ymax": 100}}]}
response = requests.post(f"{API_URL.rstrip('/')}/datasets/{DATASET_ID}/files/{FILE_ID}/labels", headers=headers, json=labels, verify=VERIFY_SSL, timeout=30)
response.raise_for_status()
print(response.json())
```
