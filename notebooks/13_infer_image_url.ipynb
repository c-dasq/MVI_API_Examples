# 13. Infer an image URL

Send a public image URL to a deployed model

Before running: Set MODEL_ID and IMAGE_URL; the MVI server must be able to reach the image URL

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
MODEL_ID = '' # The model_id can be found in the URL for that model in the MVI UI
IMAGE_URL = ''

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

response = requests.get(f"{API_URL.rstrip('/')}/dlapis/{MODEL_ID}", params={"imageUrl": IMAGE_URL}, headers=headers, verify=VERIFY_SSL, timeout=120)
response.raise_for_status()
print(response.json())
```
