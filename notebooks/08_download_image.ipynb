# 08. Download an image

Download a dataset file to your computer

Before running: Set DATASET_ID and FILE_ID; the output file may be overwritten

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_ID = '' # The dataset_id can be found in the URL for that dataset in the MVI UI
FILE_ID = '' # The file_id must be retrieved from 05_list_dataset_files
OUTPUT_PATH = 'downloaded_image.jpg'

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

response = requests.get(f"{API_URL.rstrip('/')}/datasets/{DATASET_ID}/files/{FILE_ID}/download", headers=headers, verify=VERIFY_SSL, timeout=60)
response.raise_for_status()
with open(OUTPUT_PATH, "wb") as image_file:
    image_file.write(response.content)
print("Saved:", OUTPUT_PATH)
```
