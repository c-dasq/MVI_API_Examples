# 04. Create a dataset

Create a new empty dataset

Before running: this creates a dataset

Fill in `API_URL` and `API_KEY` in the next cell; set other values if needed

```python
import requests

API_URL = ""  # Example: https://your-server/api
API_KEY = ""
VERIFY_SSL = False # Set to True if you want to verify SSL certificates
DATASET_NAME = 'training_example'

# Use the API key in each request
headers = {"X-Auth-Token": API_KEY}

response = requests.post(f"{API_URL.rstrip('/')}/datasets", headers=headers, json={"name": DATASET_NAME}, verify=VERIFY_SSL, timeout=30)
response.raise_for_status()
print(response.json())
```
